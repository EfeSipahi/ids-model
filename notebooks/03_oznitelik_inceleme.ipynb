```python
import json
from pathlib import Path
import pandas as pd

PROJECT = Path.home() / "Documents" / "ids-model"

def read_jsonl(path):
    with path.open(encoding="utf-8") as file:
        return [json.loads(line) for line in file if line.strip()]

requests = read_jsonl(
    PROJECT / "data/http-mini/requests.jsonl"
)

results = read_jsonl(
    PROJECT / "data/processed/http-mini-features-v1/features.jsonl"
)

raw_by_id = {r["sample_id"]: r for r in requests}
features_by_id = {r["sample_id"]: r for r in results}

print("İstek sayısı:", len(requests))
print("Özellik kaydı sayısı:", len(results))
```

```python
sample_id = "anomalousTrafficTest.txt:987"

request = raw_by_id[sample_id]
result = features_by_id[sample_id]

print("Kaynak etiketi:", request["label_name"])
print("Metot:", request["method"])
print("Hedef:", request["request_target"])
print("\nBody:\n", request["body"])
print("\nAyrıştırma durumu:\n", result["parse_status"])

display(
    pd.Series(result["features"], name="Değer").to_frame()
)
```
